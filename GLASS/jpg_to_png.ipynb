```python
import os
from PIL import Image

# Set your dataset root directory
root_dir = "/shared/ssd_30T/NO_WZ/DroneModel/GLASS/datasets/mvtec/thermal_image"

# Traverse the directory tree
for subdir, _, files in os.walk(root_dir):
    for file in files:
        if file.lower().endswith((".jpg", ".jpeg")):
            jpg_path = os.path.join(subdir, file)
            png_path = os.path.splitext(jpg_path)[0] + ".png"

            # Convert and save as PNG
            try:
                with Image.open(jpg_path) as im:
                    im.save(png_path, "PNG")
                os.remove(jpg_path)  # Optional: delete original JPG
                print(f"Converted: {jpg_path} -> {png_path}")
            except Exception as e:
                print(f"Failed to convert {jpg_path}: {e}")
```
